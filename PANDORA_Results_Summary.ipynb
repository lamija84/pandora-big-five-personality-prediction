# PANDORA: recorded experimental results

This notebook is a compact index of the results used in the revised study on Big Five prediction from Reddit text. It is intended for reading, not model training. Values are transcribed from saved outputs in `PANDORA_Final_Organized_Notebook.ipynb`; the source cells are identified below so the numbers can be checked against the executed analyses.

The primary evaluation used 1,094 training authors and 274 held-out test authors. Personality percentiles were normalized to a 0–1 scale. Lower MAE indicates better prediction. Supplementary analyses use the same author split but can have different training configurations; their numbers should not be substituted for the primary comparison.

## 6. Runtime snapshot and interpretation notes

The final Colab session reported Python 3.13.16, scikit-learn 1.6.1, XGBoost 3.4.1, NumPy 2.4.6, pandas 2.2.3, and SciPy 1.16.3. This is a **final-session snapshot**, not proof that every preceding experiment ran under those exact versions.

The primary selected models showed modest MAE differences relative to constant baselines. The recorded paired bootstrap and Wilcoxon comparisons did not establish statistically significant improvements over the mean or median baseline at the nominal 0.05 level. The ablation results should not be interpreted as significance-tested model comparisons.

This notebook does not contain raw Reddit comments, author IDs, model files, or credentials. For the full execution history, consult the main experimental notebook. The older exploratory analysis is preserved separately and should not be mixed with the final author-disjoint results.
